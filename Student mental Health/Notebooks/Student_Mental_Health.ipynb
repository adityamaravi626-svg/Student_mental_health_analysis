```python
missing = df_clean.isnull().sum()
missing = missing[missing>0]
missing.sort_values(ascending =False)
```

```python
missing_percentage = (
    df_clean.isnull().sum()/len(df_clean)) *100
missing_percentage.sort_values(ascending=False)
```

```python
print("Duplicates Before:",df_clean.duplicated().sum())
df_clean.drop_duplicates(inplace=True)
print("Duplicates After:",df_clean.duplicated().sum())
```

```python
df_clean.dtypes
```

```python
df_clean["What is your CGPA?"] = df_clean["What is your CGPA?"].astype(str).str.split(' ').str[0].astype(float)
```

```python
df_clean.columns = (
    df_clean.columns.str.strip().str.lower().str.replace(" ","_")
)
```

```python
text_columns = df_clean.select_dtypes(include = "object").columns
for col in text_columns:
  df_clean[col] = df_clean[col].str.strip()
```

```python
df_clean["choose_your_gender"]= (
    df_clean["choose_your_gender"].replace({
        "male":"Male",
        "MALE": "Male",
        "FEMALE":"Female",
        "female":"Female"
    })
)
```

```python
numeric_cols = df_clean.select_dtypes(include = np.number).columns
for col in numeric_cols:
  df_clean[col].fillna(df_clean[col].median(), inplace = True)
```

```python
categorical_cols = df_clean.select_dtypes(include="object").columns

for col in categorical_cols:
    df_clean[col].fillna(df_clean[col].mode()[0], inplace=True)
```

```python
df_clean.isnull().sum()
```

```python
for col in categorical_cols:
    print(col)
    print(df_clean[col].unique())
    print("----------------")
```

```python
numeric_cols = df_clean.select_dtypes(include=np.number).columns

df_clean[numeric_cols].boxplot(figsize=(15,6))

plt.xticks(rotation=90)

plt.show()
```

```python
Q1 = df_clean["what_is_your_cgpa?"].quantile(0.25)
Q3 = df_clean["what_is_your_cgpa?"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

df_clean = df_clean[
    (df_clean["what_is_your_cgpa?"] >= lower) &
    (df_clean["what_is_your_cgpa?"] <= upper)
]
```
